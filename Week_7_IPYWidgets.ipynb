```python
import ipywidgets
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
```

```python
np.random.seed(42)
x = np.random.uniform(0, 5, size=100)
noise = np.random.normal(size=100)
```

```python
def plot_fct(w=1):
    y = np.log(x) + w*noise
    plt.scatter(x, y)

plot_fct()
```

```python
ipywidgets.interact(plot_fct, w=(0, 5, 0.5))
```

```python
from sklearn.datasets import make_moons
```

```python
def plot_moons(samples=200, noise=0):
    moons = make_moons(n_samples=samples, noise=noise)
    x, y = moons[0], moons[1]
    plt.scatter(x[:, 0], x[:, 1], c=y)
```

```python
ipywidgets.interact(plot_moons, samples=[200, 500, 1000, 2000], noise=(0, 2, 0.25))
```

```python
def plot_sin(start=0, end=30, factor=1, grid=False, plot_cos=False):
    x = np.linspace(start, end, 100)
    y = np.sin(x) * factor
    plt.plot(x, y)
    if plot_cos:
        y = np.cos(x)
        plt.plot(x, y)
```
